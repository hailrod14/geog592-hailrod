# Part 5: Check the Input Data

Now improve `calculate_distance()` so that it verifies the input data before calculating the distance.

Your function should check that:

1. `point1` is a list;
2. `point2` is a list;
3. each coordinate list contains exactly two values;
4. every coordinate value has a datatype of either `int` or `float`.

For example, your function should recognize these as numeric values:

```python
type(680100) == int
type(680100.5) == float
```

If the input is not valid:

- print a useful error message;
- use `return` to stop the function before attempting the distance calculation.

````python
# Your answer to Part #5 using code: 
locations = ["A", "B", "C", "D"]

eastings = [
    680100,
    680450,
    681000,
    680300
]

northings = [
    3975200,
    3975450,
    3976100,
    3975650
]

def calculate_distance(point1, point2):
    #1 point1 must be a list
    if not isinstance(point1, list):
        print("Error: point1 is not a list")
        return 

#2 point2 must be a list
    if not isinstance(point2, list):
        print("Error: point2 is not a list")
        return

#3 each list must contain exactly 2 values
    if len(point1) != 2:
        print("Error: point1 does not contain exactly 2 values")
        return  

#4 every value must be an int or a float
    for value in point1 + point2:
        if type(value) != int and type(value) != float:
            print("Error: value is not an int or a float")
            return

  # all checks passed — calculate the distance
    easting_diff = point2[0] - point1[0]
    northing_diff = point2[1] - point1[1]
    distance = (easting_diff ** 2 + northing_diff ** 2) ** 0.5
    return distance
````
