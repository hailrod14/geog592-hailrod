## Moving to NumPy arrays

```python
import numpy as np

points = np.array([
    [100, 200],
    [140, 230],
    [180, 260],
    [120, 280]
])
```

**14. What does `points.shape` return, and what do the two numbers mean?**

*My answer:* (4, 2). 4 is rows and 2 is columns.

```python
points.shape
```

**15. What does `points[0]` return?**

*My answer:* 100, 200

```python
points[0]
```

**16. What does `points[0, 1]` return?**

*My answer:* 200

```python
points[0, 1]
```

**17. What does `points[:, 0]` return? Explain the colon.**

*My answer:* 100, 140, 180, 120. `:` means all rows, so this returns column 0 (x values).

```python
points[:, 0]
```

**18. What does `points[:, 1]` return?**

*My answer:* 200, 230, 260, 280

```python
points[:, 1]
```

**19. Write NumPy indexing code that returns the second and third rows.**

*My answer:* `points[1:3]`

```python
points[1:3]
```

**20. Compare `points[2][1]` with `points[2, 1]`.**

*My answer:* `[2][1]` takes 2 steps, pulling out the 3rd row then indexing it for the y value; `[2, 1]` goes straight to row 2, column 1. Both pull 260.

```python
print(points[2][1])
print(points[2, 1])
```

## np.where() and classification

```python
elevation = np.array([240, 510, 430, 680])
classes = np.where(elevation >= 500, 1, 0)
```

**26. What values are stored in `classes[:]`?**

**27. In your own words, explain the three arguments in `np.where(condition, value_if_true, value_if_false)`.**
